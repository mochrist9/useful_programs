## Sumowanie ciągu geometrycznego

Suma pierwszych $n$ wyrazów skończonego ciągu geometrycznego wyraża się wzorem:

$$
S_n = a_1 \cdot \frac{1 - q^n}{1 - q}, \quad q \neq 1
$$

```python
import matplotlib
# Ustawienie backendu 'Agg', aby działało w środowiskach tekstowych/chmurowych
matplotlib.use('Agg') 
import matplotlib.pyplot as plt

# Definiujemy dane dla k i y
k_values = list(range(1, 21))
y_values = [k**2 for k in k_values]

# Tworzenie wykresu
plt.figure(figsize=(8, 5))
plt.plot(k_values, y_values, marker='o', linestyle='-', color='blue', linewidth=2)
plt.title('Wykres funkcji y = k^2 dla k = 1, ..., 20')
plt.xlabel('k')
plt.ylabel('y')
plt.grid(True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import imageio_ffmpeg

# Automatyczne wskazanie ścieżki do ffmpeg dostarczonego przez Python
plt.rcParams['animation.ffmpeg_path'] = imageio_ffmpeg.get_ffmpeg_exe()

# Parametry wahadła
L = 1.0       # długość nitki
g = 9.81      # grawitacja
theta0 = np.pi / 4  # początkowy kąt wychylenia

dt = 0.05
t = np.arange(0, 5, dt)
theta = theta0 * np.cos(np.sqrt(g / L) * t)

x = L * np.sin(theta)
y = -L * np.cos(theta)

fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 0.5)
ax.set_aspect('equal')
ax.grid(True)

line, = ax.plot([], [], 'o-', lw=3, markersize=12, color='red')
time_text = ax.text(0.05, 0.9, '', transform=ax.transAxes)

def init():
    line.set_data([], [])
    time_text.set_text('')
    return line, time_text

def animate(i):
    thisx = [0, x[i]]
    thisy = [0, y[i]]
    line.set_data(thisx, thisy)
    time_text.set_text(f'Czas: {t[i]:.2f} s')
    return line, time_text

ani = animation.FuncAnimation(fig, animate, frames=len(t), init_func=init, blit=True, interval=dt*1000)

# Zapis do video.mp4
writer = animation.FFMpegWriter(fps=20, metadata=dict(artist='User'), bitrate=1800)
ani.save('video.mp4', writer=writer)
plt.close()
print("Zapisano animację do pliku 'video.mp4'")

# Osadzenie wideo w notebooku
display(HTML('<video controls autoplay><source src="video.mp4" type="video/mp4"></video>'))
```

```python
# Zadanie 11: Zmodyfikowana krzywa parametryczna (Spirala / Helisa 3D)
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Modyfikacja równań parametrycznych – tworzymy spiralę (helisę)
t = np.linspace(0, 4 * np.pi, 500)  # zakres od 0 do 4*pi dla dłuższego zwoju
x = np.cos(t)
y = np.sin(t)
z = t / (2 * np.pi)                 # liniowy wzrost wzdłuż osi z

# Tworzenie wykresu 3D
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

# Rysowanie zmodyfikowanej krzywej
ax.plot(x, y, z, label="Zmodyfikowana krzywa: Helisa", color="purple", linewidth=2.5)

# Konfiguracja tytułu i etykiet osi
ax.set_title("Zadanie 11: 3D Parametric Helix", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()

# [WAŻNE] Zapisanie wykresu do pliku przed wywołaniem plt.show()
plt.savefig("parametric_curve.png", dpi=300, bbox_inches='tight')

# Wyświetlenie wykresu w notebooku
plt.show()

print("Sukces! Wykres został wyświetlony oraz zapisany jako 'parametric_curve.png'.")
```
